### **3.2 Hitung masalah**

Hasil hitungan ini disimpan di variabel (`n_...`) s

```python
# Daftar kolom teks (dipakai berulang)
kolom_teks = ['OPERATOR_NAME', 'SUBUNIT', 'COAL_METAL_IND', 'CLASSIFICATION', 'ACCIDENT_TYPE',
              'OCCUPATION', 'ACTIVITY', 'NATURE_INJURY', 'INJ_BODY_PART', 'DEGREE_INJURY']

# 1. Nilai kosong tersamar: teks "NO VALUE FOUND" yang sebenarnya berarti tidak ada data
kolom_tersamar = ['OCCUPATION', 'ACTIVITY', 'NATURE_INJURY', 'INJ_BODY_PART']
n_placeholder = (raw_selected[kolom_tersamar].apply(lambda s: s.str.strip()) == 'NO VALUE FOUND').sum().sum()

# 2. Spasi berlebih di awal/akhir teks
n_spasi = 0
for kolom in kolom_teks:
    isi = raw_selected[kolom].dropna()
    n_spasi = n_spasi + (isi != isi.str.strip()).sum()

# 3. Jam kejadian tidak valid (di atas 23:59, atau menit di atas 59, misal 9999)
jam = pd.to_numeric(raw_selected['ACCIDENT_TIME'], errors='coerce')
n_jam = ((jam > 2359) | (jam % 100 > 59)).sum()

# 4. Pengalaman kerja tidak logis (pengalaman di pekerjaan/tambang > total pengalaman)
n_logika = ((raw_selected['JOB_EXPER'] > raw_selected['TOT_EXPER']) | (raw_selected['MINE_EXPER'] > raw_selected['TOT_EXPER'])).sum()

# 5. DAYS_LOST kosong --> menghitung hari kerja yang hilang akibat cedera
n_lost_kosong = raw_selected['DAYS_LOST'].isnull().sum()

# 6. OPERATOR_ID yang punya lebih dari satu penulisan nama
n_operator = (raw_selected.groupby('OPERATOR_ID')['OPERATOR_NAME'].nunique() > 1).sum()

# 7. Nilai ekstrem DAYS_LOST (lebih dari 365 hari)
n_lebih_365 = (raw_selected['DAYS_LOST'] > 365).sum()

# 8. Operator kosong
n_operator_kosong = raw_selected['OPERATOR_ID'].isnull().sum()
```



### **3.7 Jam kerja tidak valid**


`ACCIDENT_TIME` = 9999 adalah kode "tidak diketahui", bukan jam 99:99. Kalau dipakai, analisis "jam paling rawan" salah total. Nilainya diubah ke NaN, lalu dibuat kolom `ACCIDENT_HOUR` (0–23) untuk pertanyaan bisnis tentang jam kejadian.



```python
# 1. Konversi ke numeric (teks aneh akan otomatis jadi NaN)
jam = pd.to_numeric(raw_selected["ACCIDENT_TIME"], errors="coerce")

# 2. Definisikan kondisi tidak valid
# Edge case: Jam tidak boleh negatif, tidak boleh > 2359, dan menit tidak boleh > 59
tidak_valid = (jam < 0) | (jam > 2359) | (jam % 100 > 59)

# 3. Hitung jam kejadian (0-23)
raw_selected["ACCIDENT_HOUR"] = jam // 100

# 4. BEST PRACTICE: Ubah kedua kolom sekaligus dalam 1 baris saja (lebih cepat)
raw_selected.loc[tidak_valid, ["ACCIDENT_TIME", "ACCIDENT_HOUR"]] = np.nan

# 5. Cek hasil
print("Jam tidak valid dijadikan NaN:", tidak_valid.sum())
print(raw_selected["ACCIDENT_HOUR"].value_counts().sort_index().head(5))
```



## **4. Outliers (nilai ekstrem)**

kolom DAYS_LOST = hari kerja yang hilang akibat kecelakaan, sebagian besar kecelakaan di lapangan berupa luka ringan biasa dan pekerja dapat langsung kembali bekerja di hari itu juha (DAYS_LOST = 0)

```python
# Ambil kasus yang punya hari hilang (lebih dari 0)
lost_positif = raw_selected[raw_selected['DAYS_LOST'] > 0]['DAYS_LOST']

# Hitung kuartil 1 dan kuartil 3
Q1 = lost_positif.quantile(0.25)
Q3 = lost_positif.quantile(0.75)

# Hitung IQR dan batas atas (batas bawah tidak relevan karena hari hilang tidak mungkin negatif)
IQR = Q3 - Q1
batas_atas = Q3 + (1.5 * IQR)

print(f"Q1          : {Q1}")
print(f"Q3          : {Q3}")
print(f"IQR         : {IQR}")
print(f"Batas atas  : {batas_atas}")
```





1.   `outlier_iqr`: Menangkap semua kasus yang secara matematika jauh di atas rata-rata kasus cedera biasa.
2.   `lebih_365`: Ada 365 hari dalam setahun, maka kehilangan hari kerja di atas 1 tahun adalah kondisi luar biasa. Apakah ini human error (misal diisi kode error 9999), atau memang kecelakaan fatal? harus dianalisis lanjut



```python
# Saring baris yang DAYS_LOST-nya di atas batas atas IQR
outlier_iqr = raw_selected[raw_selected['DAYS_LOST'] > batas_atas]
print("Outlier menurut IQR      :", len(outlier_iqr), "dari", len(lost_positif), "kasus dengan hari hilang")

# Cek logika bisnis: satu tahun = 365 hari
lebih_365 = raw_selected[raw_selected['DAYS_LOST'] > 365]
print("Kasus lebih dari 365 hari:", len(lebih_365))

# Outlier ini kategori cedera apa?
lebih_365['DEGREE_INJURY'].value_counts()
```

```python
# Kolom penanda outlier: isi awal 0, lalu 1 untuk yang di atas batas IQR
raw_selected['OUTLIER_DAYS_LOST'] = 0
raw_selected.loc[raw_selected['DAYS_LOST'] > batas_atas, 'OUTLIER_DAYS_LOST'] = 1

raw_selected['OUTLIER_DAYS_LOST'].value_counts()
```

```python

# Path file di Google Drive
txt_path = '/content/drive/MyDrive/Final_Project/ContractorProdYearly.txt'

# Baca file join_raw
join_raw = pd.read_csv(
    txt_path,
    sep='|',
    quotechar='"',
    encoding='latin-1',
    low_memory=False
)

print(f"Data join_raw berhasil dimuat: {join_raw.shape[0]} baris, {join_raw.shape[1]} kolom.")
```

```python
# 1. Bersihkan string ID dan Komoditas
join_raw['CONTRACTOR_ID'] = join_raw['CONTRACTOR_ID'].astype(str).str.strip()
join_raw['COAL_METAL_IND'] = join_raw['COAL_METAL_IND'].astype(str).str.strip().str.upper()

# 2. Standarisasi SUBUNIT_CD ke 2 digit (misal '1' -> '01')
join_raw['SUBUNIT_CD'] = join_raw['SUBUNIT_CD'].astype(str).str.strip().str.zfill(2)

# 3. Ubah Tahun ke Integer
join_raw['CALENDAR_YR'] = pd.to_numeric(join_raw['CALENDAR_YR'], errors='coerce').astype('Int64')

# 4. Bersihkan kolom numerik (Jam Kerja & Jumlah Karyawan)
join_raw['ANNUAL_HOURS'] = pd.to_numeric(join_raw['ANNUAL_HOURS'], errors='coerce').fillna(0)
join_raw['AVG_EMPLOYEE_CNT'] = pd.to_numeric(join_raw['AVG_EMPLOYEE_CNT'], errors='coerce').fillna(0)

# Cek hasil pembersihan
join_raw[['CONTRACTOR_ID', 'CALENDAR_YR', 'SUBUNIT_CD', 'COAL_METAL_IND', 'ANNUAL_HOURS']].head()
```

```python
# Buat salinan agar data asli tidak tertimpa rusak
accident_clean = raw_selected.copy()

# 1. Bersihkan CONTRACTOR_ID (ganti 'nan', string kosong dengan NaN sesungguhnya)
accident_clean['CONTRACTOR_ID'] = accident_clean['CONTRACTOR_ID'].astype(str).str.strip()
accident_clean['CONTRACTOR_ID'] = accident_clean['CONTRACTOR_ID'].replace(['nan', 'None', '', 'NO VALUE FOUND'], np.nan)

# 2. Standarisasi SUBUNIT_CD ke 2 digit persis seperti join_raw
accident_clean['SUBUNIT_CD'] = accident_clean['SUBUNIT_CD'].astype(str).str.strip().str.zfill(2)

# 3. Standarisasi COAL_METAL_IND dan CAL_YR
accident_clean['COAL_METAL_IND'] = accident_clean['COAL_METAL_IND'].astype(str).str.strip().str.upper()
accident_clean['CAL_YR'] = pd.to_numeric(accident_clean['CAL_YR'], errors='coerce').astype('Int64')

print("Format kunci raw_filtered berhasil diselaraskan!")
```

```python
# 1. Pilih kolom-kolom penting dari join_raw yang mau kita bawa
lookup_cols = [
    'CONTRACTOR_ID', 'CALENDAR_YR', 'SUBUNIT_CD', 'COAL_METAL_IND',
    'CONTRACTOR_NAME', 'ANNUAL_HOURS', 'AVG_EMPLOYEE_CNT'
]

# 2. Hapus duplikasi kunci di join_raw untuk mencegah baris berlipat ganda
prod_lookup = join_raw[lookup_cols].drop_duplicates(
    subset=['CONTRACTOR_ID', 'CALENDAR_YR', 'SUBUNIT_CD', 'COAL_METAL_IND']
)

# 3. Gabungkan data (Merge)
merged_data = pd.merge(
    accident_clean,
    prod_lookup,
    left_on=['CONTRACTOR_ID', 'CAL_YR', 'SUBUNIT_CD', 'COAL_METAL_IND'],
    right_on=['CONTRACTOR_ID', 'CALENDAR_YR', 'SUBUNIT_CD', 'COAL_METAL_IND'],
    how='left'
)

# Hapus kolom CALENDAR_YR karena nilainya sudah sama dengan CAL_YR
merged_data.drop(columns=['CALENDAR_YR'], inplace=True)

print(f"Penggabungan selesai!")
print(f"Total baris awal (raw_selected): {len(raw_selected)}")
print(f"Total baris setelah join        : {len(merged_data)}")
```

```python
df_final = merged_data.copy()
# 1. FILTER TAHUN (Pertanyaan Bisnis 1: 2015 - 2025)
df_final['CAL_YR'] = pd.to_numeric(df_final['CAL_YR'], errors='coerce')
df_final = df_final[(df_final['CAL_YR'] >= 2015) & (df_final['CAL_YR'] <= 2025)].copy()

# 2. TOTAL HARI KERJA HILANG
df_final['DAYS_LOST'] = pd.to_numeric(df_final['DAYS_LOST'], errors='coerce').fillna(0)
df_final['DAYS_RESTRICT'] = pd.to_numeric(df_final['DAYS_RESTRICT'], errors='coerce').fillna(0)

# Total hari hilang = hari tidak masuk kerja + hari tugas terbatas
df_final['TOTAL_DAYS_LOST'] = df_final['DAYS_LOST'] + df_final['DAYS_RESTRICT']

# 3. TINGKAT KEPARAHAN (Pertanyaan Bisnis 2, 4, & 5)
# Definisi Kasus Fatal (Kode MSHA 01 = Fatal)
is_fatal = (
    df_final['DEGREE_INJURY_CD'].astype(str).str.strip().isin(['01', '1']) |
    df_final['DEGREE_INJURY'].astype(str).str.upper().str.contains('FATAL')
)
df_final['IS_FATAL'] = np.where(is_fatal, 1, 0)

# Definisi Kasus Berat: Fatal ATAU Cacat Permanen ATAU Total Hari Hilang >= 14 hari
# (14 hari adalah threshold standar industri tambang untuk Lost-Time Injury signifikan)
is_severe = (
    (df_final['IS_FATAL'] == 1) |
    (df_final['DEGREE_INJURY_CD'].astype(str).str.strip().isin(['02', '03', '2', '3'])) |
    (df_final['TOTAL_DAYS_LOST'] >= 14)
)
df_final['IS_SEVERE'] = np.where(is_severe, 1, 0)

# 4.AREA KERJA & KOMODITAS (Pertanyaan Bisnis 2)
# Mapping Subunit MSHA menjadi Area Kerja umum yang mudah dibaca
def map_work_area(code):
    c = str(code).zfill(2)
    if c in ['01', '02']:
        return 'Underground'
    elif c in ['03', '04', '05', '06']:
        return 'Open Pit / Surface'
    elif c in ['30', '31', '32']:
        return 'Mill / Prep Plant'
    elif c in ['99']:
        return 'Office / Facility'
    return 'Other'

df_final['WORK_AREA'] = df_final['SUBUNIT_CD'].apply(map_work_area)

# Label Komoditas yang jelas
df_final['COMMODITY_TYPE'] = df_final['COAL_METAL_IND'].map({
    'C': 'Coal',
    'M': 'Metal & Non-Metal'
}).fillna('Other')

# 5. KELOMPOK PENGALAMAN KERJA (Pertanyaan Bisnis 3)
# Menggunakan nama kolom yang benar: MINE_EXPER
df_final['MINE_EXPER'] = pd.to_numeric(df_final['MINE_EXPER'], errors='coerce')

# Buat kelompok: < 1 Tahun vs >= 1 Tahun
df_final['EXP_GROUP'] = np.where(
    df_final['MINE_EXPER'].isna(), 'Unknown',
    np.where(df_final['MINE_EXPER'] < 1.0, '< 1 Year', '>= 1 Year')
)

# 6. WAKTU (KUARTAL & JAM KECELAKAAN) (Pertanyaan Bisnis 5)
# Parse Tanggal untuk Kuartal
df_final['ACCIDENT_DT'] = pd.to_datetime(df_final['ACCIDENT_DT'], errors='coerce')
df_final['QUARTER'] = 'Q' + df_final['ACCIDENT_DT'].dt.quarter.astype('Int64').astype(str)
df_final['QUARTER'] = df_final['QUARTER'].replace('Q<NA>', np.nan)

# Ekstrak Jam (ACCIDENT_TIME format MSHA biasanya 4 digit: '0930', '1400', atau float: 930.0)
def extract_hour(val):
    if pd.isna(val):
        return np.nan
    val_clean = str(val).split('.')[0].strip().zfill(4)
    if len(val_clean) == 4 and val_clean.isdigit():
        hour = int(val_clean[:2])
        if 0 <= hour <= 23:
            return hour
    return np.nan

df_final['ACCIDENT_HOUR'] = df_final['ACCIDENT_TIME'].apply(extract_hour).astype('Int64')

print("Ukuran dataset:", df_final.shape)
```

```python
# Dictionary FIPS Code Tambang ke Singkatan Negara Bagian AS (MSHA Standard)
fips_to_state = {
    1: 'AL', 2: 'AZ', 4: 'CA', 5: 'CO', 6: 'CT', 8: 'FL', 9: 'GA',
    10: 'ID', 11: 'IL', 12: 'IN', 13: 'IA', 14: 'KS', 15: 'KY', 16: 'LA',
    17: 'ME', 18: 'MD', 19: 'MA', 20: 'MI', 21: 'MN', 22: 'MS', 23: 'MO',
    24: 'MT', 25: 'NE', 26: 'NV', 27: 'NH', 28: 'NJ', 29: 'NM', 30: 'NY',
    31: 'NC', 32: 'ND', 33: 'OH', 34: 'OK', 35: 'OR', 36: 'PA', 37: 'RI',
    38: 'SC', 39: 'SD', 40: 'TN', 41: 'TX', 42: 'UT', 43: 'VT', 44: 'VA',
    45: 'WA', 46: 'WV', 47: 'WI', 48: 'WY', 50: 'AK', 52: 'PR'
}

# 1. Tambahkan Kolom STATE untuk Peta
df_final['STATE'] = pd.to_numeric(df_final['FIPS_STATE_CD'], errors='coerce').map(fips_to_state).fillna('Other')

# 2. Tambahkan Kolom Rentang Jam Teks
def format_hour(val):
    if pd.isna(val) or val == '':
        return 'Unknown'
    return f"{int(val):02d}:00 - {int(val):02d}:59"

df_final['HOUR_INTERVAL'] = df_final['ACCIDENT_HOUR'].apply(format_hour)

# 3. Buat Label Teks untuk Dropdown Filter Looker Studio
df_final['SEVERE_STATUS'] = df_final['IS_SEVERE'].map({1: 'Severe', 0: 'Non-Severe'})
df_final['FATAL_STATUS'] = df_final['IS_FATAL'].map({1: 'Fatal', 0: 'Non-Fatal'})

# 4. Pastikan Tanggal berformat YYYY-MM-DD
df_final['ACCIDENT_DT'] = pd.to_datetime(df_final['ACCIDENT_DT'], errors='coerce').dt.strftime('%Y-%m-%d')

# 5. Export ke CSV Final Siap Pakai Looker Studio
export_final_path = '/content/drive/MyDrive/Final_Project/Data_Ready.csv'

df_final.to_csv(export_final_path, index=False, encoding='utf-8')

```
